# Ujian Kecerdasan Buatan : Klasifikasi

## 1. Pengolahan Data

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('data_pasien.csv')
```

```python
df.info()
```

```python
df.head()
```

```python
print("5 data terakhir")
df.tail()
```

## 2. Pembersihan Data

```python
df.isnull().sum()
```

```python
mapping = {'Ya': 1, 'Tidak': 0}
for col in ["Merokok", "Olahraga", "Penyakit"]:
    df[col] = df[col].astype(str).str.strip().map(mapping)
print(df)
```

## 3. Pemrosesan Data

```python
print("pisahkan fitur dan LABEL")
X = df.drop(columns=['ID', 'Penyakit'])
y = df['Penyakit']
```

```python
print(" Split data 90:10")
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.1, random_state=42
)
```

```python
print(" Normalisasi data numerik")
num_cols = ['Usia', 'IMT', 'TekananDarah']
scaler = StandardScaler()
X_train[num_cols] = scaler.fit_transform(X_train[num_cols])
X_test[num_cols] = scaler.transform(X_test[num_cols])
```

## 4. Klasifikasi

```python
print(" Desision Tree Classifier")
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print("Akurasi Model Decision Tree: {:.2f}%".format(accuracy * 100))
```
